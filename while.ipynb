```python
#Exercicio 1
contador = 10

while contador >= 0:
  print("Contagem: ", contador)
  contador -= 1
print("Fim da contagem!")
```

```python
#Exercicio 2
contador = 0

while contador <= 20:
  print("Contagem: ", contador)
  contador % 2 == 0
  contador +=2
print("Fim da contagem!")
```

```python
#Exercicio 3
soma = 0
numero = 0
while numero != -5:
  numero = int(input("Digite um numero: "))
  if numero != -1:
    soma += numero
  print("a soma dos numeros:", soma)
```

```python
#Exercicio 4
senha_corrta = "cofre2025"
tentativa = ""

while tentativa != senha_corrta:
  tentativa = input("Digite a código do cofre: ")
print("Cofre aberto com sucesso!")
```
